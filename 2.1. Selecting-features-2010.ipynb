```python
# FEATURES HOUSING

# Define the list of columns to keep
columns_to_keep = [
    "Household ID",
    "Province",
    "District (also stratum for SE calculation)",
    "Urban/rural 2002",
    "Quintile",
    "Poverty status",
    "Final weight of household",
    "Cluster",
    "Type of habitat",
    "Type of dwelling",
    "Number of rooms occupied by household",
    "Number of bed rooms for household",
    "Any rooms shared with other households",
    "Owner of the dwelling",
    "Main source of drinking water",
    "Distance to the nearest drinking water source (meters)",
    "Amount paid to private water vendor last 7 days",
    "Main source of home lighting",
    "Amount paid on electricity in last 4 weeks",
    "Primary source of cooking fuel",
    "Mode of rubbish disposal",
    "Type of toilet household use",
    "Household own a working telephone",
    "Internet connection at home",
    "Main construction material of exterior wall",
    "Main roofing material for dwelling",
    "Main floor material for dwelling",
    "Area of the floor for dwelling-outside",
    "Current occupancy status",
    "Main source of info on environmental issues"
]

# Create the new DataFrame
features_housing = S05ABCD_housing[columns_to_keep]
```

```python
# DEMOGRAPHIC FEATURES

# Filter for rows where the relation is 'Household head (HH)'
household_heads = S01_hhmembers[S01_hhmembers["Relation with the head of household"] == "Household head (HH)"]
```

```python
# Define the columns to keep
columns_to_keep = [
    "Household ID",
    "Sex",
    "Age (Years)",
    "Marital status",
    "Place of birth",
    "Ever been to school",
    "Attended school in the last 12 months",
    "Highest class succesfully attained in school",
    "Highest school diploma or certificate obtained",
    "Able to read a simple written note",
    "Able to write a simple letter",
    "Language able to write",
    "Able to do a written calculation",
    "Health insurance",
    "Best writting language",
    "Ever used a computer",
    "Confidence to use computer",
    "Purchase of medicines or medical supplies in the last 2 weeks"
]

# Subset the dataframe
household_heads = household_heads[columns_to_keep]
```

```python
# Impute 'No' in 'Confidence to use computer' if 'Ever used a computer' is 'No'
household_heads.loc[
    household_heads["Ever used a computer"] == "No",
    "Confidence to use computer"
] = "No"
```

```python
# Calculate household size
household_size = (
    S01_hhmembers.groupby("Household ID")
    .size()
    .reset_index(name="Household size")
)
```

```python
# Filter and count number of kids per household
number_of_kids = (
    S01_hhmembers[S01_hhmembers["Relation with the head of household"] == "Son or daughter of HH"]
    .groupby("Household ID")
    .size()
    .reset_index(name="Number of kids")
)
```

```python
# Merge household size
household_heads = household_heads.merge(household_size, on="Household ID", how="left")

# Merge number of kids (fill NaN with 0 because some households may not have kids)
household_heads = household_heads.merge(number_of_kids, on="Household ID", how="left")
household_heads["Number of kids"] = household_heads["Number of kids"].fillna(0).astype(int)
```

```python
# FEATURES LAND

# Make a copy to avoid modifying the original data directly
S08C_parcels_cleaned = S08C_parcels.copy()

# Convert the Surface area to numeric (this handles categorical or string values)
S08C_parcels_cleaned["Surface area"] = pd.to_numeric(S08C_parcels_cleaned["Surface area"], errors='coerce')

# Now group and sum
household_land_area = (
    S08C_parcels_cleaned
    .groupby("Household ID")["Surface area"]
    .sum()
    .reset_index()
    .rename(columns={"Surface area": "Total surface area of land owned"})
)
```

```python
# Merge household_heads with total land area
final_land_df = household_heads[["Household ID"]].merge(
    household_land_area, on="Household ID", how="left"
)

# Create the yes/no column based on whether land was reported
final_land_df["Household reported owning any land"] = final_land_df[
    "Total surface area of land owned"
].notna().map({True: "yes", False: "no"})
```

```python
# FEATURES HARVEST

# Make a copy to avoid modifying the original data
S08D_croplarge_cleaned = S08D_croplarge.copy()

# Convert the harvest value column to numeric (in case it's categorical or string)
S08D_croplarge_cleaned["Value of entire harvest"] = pd.to_numeric(
    S08D_croplarge_cleaned["Value of entire harvest"], errors='coerce'
)

# Group by Household ID and sum the harvest value
household_crop_value = (
    S08D_croplarge_cleaned
    .groupby("Household ID")["Value of entire harvest"]
    .sum()
    .reset_index()
    .rename(columns={"Value of entire harvest": "Total value of large crop harvest"})
)
```

```python
# Merge household_heads with crop harvest values
final_harvest_df = household_heads[["Household ID"]].merge(
    household_crop_value, on="Household ID", how="left"
)

# Create the yes/no column based on whether Total value of large crop harvest is missing
final_harvest_df["Household harvested any large crops in last 12 months"] = final_harvest_df[
    "Total value of large crop harvest"
].notna().map({True: "yes", False: "no"})
```

```python
# FEATURES LIVESTOCK

# Make a copy of the livestock data
S08A1_livestock_cleaned = S08A1_livestock.copy()

# Convert the unit value column to numeric in case it's stored as a category or string
S08A1_livestock_cleaned["Current unit value of the livestock"] = pd.to_numeric(
    S08A1_livestock_cleaned["Current unit value of the livestock"], errors='coerce'
)

# Group by Household ID and sum the values
household_livestock_value = (
    S08A1_livestock_cleaned
    .groupby("Household ID")["Current unit value of the livestock"]
    .sum()
    .reset_index()
    .rename(columns={"Current unit value of the livestock": "Total value of livestock"})
)
```

```python
# Merge household_heads with livestock values
final_livestock_df = household_heads[["Household ID"]].merge(
    household_livestock_value, on="Household ID", how="left"
)

# Create the yes/no column based on whether Total value of livestock is missing
final_livestock_df["Household raised any livestock in last 12 months"] = final_livestock_df[
    "Total value of livestock"
].notna().map({True: "yes", False: "no"})
```

**Final Merge**

```python
# Start with features_housing as the base
all_features = features_housing.copy()

# Merge in household_heads
all_features = all_features.merge(household_heads, on="Household ID", how="left")

# Merge in final_land_df
all_features = all_features.merge(final_land_df, on="Household ID", how="left")

# Merge in final_harvest_df
all_features = all_features.merge(final_harvest_df, on="Household ID", how="left")

# Merge in final_livestock_df
all_features = all_features.merge(final_livestock_df, on="Household ID", how="left")
```

**Adjust telephone column**

```python
all_features["Household own a working telephone"].unique()
```

```python
all_features["Household own a working telephone"] = all_features["Household own a working telephone"].replace({
    "Yes, mobile phone": "yes",
    "Yes, both land line and mobile phone": "yes",
    "Yes, landline": "no",
    "None": "no"
})

all_features.rename(columns={"Household own a working telephone": "Has a mobile telephone"}, inplace=True)
```

```python
# List of columns to drop
columns_to_drop = [
    "Any rooms shared with other households",
    "Area of the floor for dwelling-outside",
    "Language able to write",
    "Best writting language",
    "Ever used a computer",
    "Purchase of medicines or medical supplies in the last 2 weeks",
    "Poverty status",
]

# Drop columns (ignore if they are missing to avoid errors)
all_features = all_features.drop(columns=columns_to_drop, errors='ignore')
```
