# Selecting features (2016)

**From cs_s0_s5_household.dta dataframe**

- s5aq2: Type of dwelling
- s5aq5: Number of rooms occupied by HH
- s5aq10: Owner of house/dwelling 
- s5aq11: Current occupancy status
- s5cq1: Type main source of water 
- s5cq4: Metres to nearest drinking water source
- s5cq3: Type nearest source of drinking water
- s5cq11: Amount paid to private water vendor/neighbour in last 7 days
- s5cq16: Main source of lighting in home 
- s5cq17: Amount paid for electricity in last 4 weeks 
- s5cq18: Primary source of cooking fuel
- s5cq19: Mode of rubbish/garbage disposal
- s5cq20: Type of toilet 
- s5cq25: Main source of info on environmental issues
- s5cq26: Have Internet connection 
- s5dq1: Main construction material of exterior wall 
- s5dq2: Main roofing material for main dwelling
- s5dq3: Main floor material for dwelling 
- s5dq4: Floor area of the dwelling (square metres) 
- Number of children (needs to be derived)

**From cs_s1_s2_s3_s4_s6a_s6e_person dataframe**
- s3q3: Main health insurance 
- s4aq1: Ever been to school 
- s4aq2: Highest class successfully completed
- s4bq3: Can read a letter or a simple note
- s4bq4: Can write a letter or a simple note 
- 4bq5: Can perform a written calculation
- s4bq6: Feels confident about using a computer 
- s4bq8: Has a mobile telephone
- s6fq4: Number of hours foraging for wood last 7 days 

**From cs_s7d_large_crop dataframe**
- s7dq11 Value received for sales of entire harvest 

**From vup_s7c_parcels.dta dataframe**
- Surface area of all land owned

```python
import pandas as pd

# Define the base path for the 2016 dataset
base_path = "/Users/gianjaeger/Desktop/2016/rwa-nisr-EICV5-data_STATA"

# Household & Person files
cs_s0_s5_household = pd.read_stata(f"{base_path}/cs_S0_S5_Household.dta")
cs_s1_s2_s3_s4_s6a_s6e_person = pd.read_stata(f"{base_path}/cs_S1_S2_S3_S4_S6A_S6E_Person.dta")

# Services and Employment
cs_s5f_access_to_services = pd.read_stata(f"{base_path}/cs_S5F_Access_to_services.dta")
cs_s6b_employement_salaried_business = pd.read_stata(f"{base_path}/cs_S6B_Employement_6C_Salaried_S6D_Business.dta")

# Livestock
cs_s7a1_livestock = pd.read_stata(f"{base_path}/cs_S7A1_livestock.dta")
cs_s7a2_livestock = pd.read_stata(f"{base_path}/cs_S7A2_livestock.dta")
cs_s7a3_livestock = pd.read_stata(f"{base_path}/cs_S7A3_livestock.dta")
cs_s7a4_livestock = pd.read_stata(f"{base_path}/cs_S7A4_livestock.dta")

# Land
cs_s7b1_land_agriculture = pd.read_stata(f"{base_path}/cs_S7B1_land_Agriculture.dta")
cs_s7b2_land_agriculture = pd.read_stata(f"{base_path}/cs_S7B2_land_Agriculture.dta")
cs_s7c_parcels = pd.read_stata(f"{base_path}/cs_S7C_parcels.dta")

# Crops
cs_s7d_large_crop = pd.read_stata(f"{base_path}/cs_S7D_large_crop.dta")
cs_s7e_small_crop = pd.read_stata(f"{base_path}/cs_S7E_small_crop.dta")

# Agriculture income and expenses
cs_s7f_income_agriculture = pd.read_stata(f"{base_path}/cs_S7F_income_agriculture.dta")
cs_s7g_expenditure_agriculture = pd.read_stata(f"{base_path}/cs_S7G_expenditure_agriculture.dta")
cs_s7h_transformation_agriculture = pd.read_stata(f"{base_path}/cs_S7H_transformation_agriculture.dta")

# Expenditures
cs_s8a1_expenditure = pd.read_stata(f"{base_path}/cs_S8A1_expenditure.dta")
cs_s8a2_expenditure = pd.read_stata(f"{base_path}/cs_S8A2_expenditure.dta")
cs_s8a3_expenditure = pd.read_stata(f"{base_path}/cs_S8A3_expenditure.dta")
cs_s8b_expenditure = pd.read_stata(f"{base_path}/cs_S8B_expenditure.dta")
cs_s8c_farming = pd.read_stata(f"{base_path}/cs_S8C_farming.dta")

# Transfers
cs_s9a_transfers_out = pd.read_stata(f"{base_path}/cs_S9A_transfers_out.dta")
cs_s9b_transfers_in = pd.read_stata(f"{base_path}/cs_S9B_transfers_in.dta")

# VUP/Ubudehe/RSSP schemes
cs_s9c_vup = pd.read_stata(f"{base_path}/cs_S9C_Vup_ubudehe_and_Rssp_schemes.dta")
cs_s9c3_vup = pd.read_stata(f"{base_path}/cs_S9C3_Vup_ubudehe_and_Rssp_schemes.dta")
cs_s9c4_vup = pd.read_stata(f"{base_path}/cs_S9C4_Vup_ubudehe_and_Rssp_schemes.dta")

# Other income and expenditure
cs_s9d_other_income = pd.read_stata(f"{base_path}/cs_S9D_other_income.dta")
cs_s9e_other_expenditure = pd.read_stata(f"{base_path}/cs_S9E_Other_expenditure.dta")

# Credits and durable goods
cs_s10a1_credits = pd.read_stata(f"{base_path}/cs_S10A1_Credits.dta")
cs_s10a2_listing_of_credits = pd.read_stata(f"{base_path}/cs_S10A2_Listing_of_credits.dta")
cs_s10b1_durable_household_goods = pd.read_stata(f"{base_path}/cs_S10B1_Durable_household_goods.dta")
cs_s10b2_durable_household_goods = pd.read_stata(f"{base_path}/cs_S10B2_Durable_household_goods.dta")

# Savings and tontine
cs_s10c1_savings = pd.read_stata(f"{base_path}/cs_S10C1_Savings.dta")
cs_s10c2_tontine = pd.read_stata(f"{base_path}/cs_S10C2_Tontine.dta")

# Poverty file
eicv5_poverty_file = pd.read_stata(f"{base_path}/EICV5_Poverty_file.dta")
```

```python
# Select and rename relevant columns from cs_s0_s5_household
features_housing = cs_s0_s5_household[[
    "hhid",
    "province",
    "district",
    "ur",
    "weight",
    "clust",
    "s5aq1",
    "s5aq2",
    "s5aq5",
    "s5aq6",
    "s5aq10",
    "s5aq11",
    "s5cq1",
    "s5cq4",
    "s5cq11",
    "s5cq16",
    "s5cq17",
    "s5cq18",
    "s5cq19",
    "s5cq20",
    "s5cq25",
    "s5cq26",
    "s5dq1",
    "s5dq2",
    "s5dq3"
]].rename(columns={
    "hhid": "Household ID",
    "province": "Province",
    "district": "District (also stratum for SE calculation)",
    "ur": "Urban/rural 2002",
    "weight": "Final weight of household",
    "clust": "Cluster",
    "s5aq1": "Type of habitat",
    "s5aq2": "Type of dwelling",
    "s5aq5": "Number of rooms occupied by household",
    "s5aq6": "Number of bed rooms for household",
    "s5aq10": "Owner of the dwelling",
    "s5aq11": "Current occupancy status",
    "s5cq1": "Main source of drinking water",
    "s5cq4": "Distance to the nearest drinking water source (meters)",
    "s5cq11": "Amount paid to private water vendor last 7 days",
    "s5cq16": "Main source of home lighting",
    "s5cq17": "Amount paid on electricity in last 4 weeks",
    "s5cq18": "Primary source of cooking fuel",
    "s5cq19": "Mode of rubbish disposal",
    "s5cq20": "Type of toilet household use",
    "s5cq25": "Main source of info on environmental issues",
    "s5cq26": "Internet connection at home",
    "s5dq1": "Main construction material of exterior wall",
    "s5dq2": "Main roofing material for dwelling",
    "s5dq3": "Main floor material for dwelling"
})
```

```python
# Select and rename relevant columns
person_features = cs_s1_s2_s3_s4_s6a_s6e_person[[
    "hhid",
    "quintile",
    "s1q1",
    "s1q2",
    "s1q3y",
    "s1q4",
    "s1q5",
    "s3q2",
    "s4aq1",
    "s4aq3",
    "s4aq2",
    "s4aq7",
    "s4bq4",
    "s4bq5",
    "s4bq8b",
    "s4bq6",
    "s4bq7"
]].rename(columns={
    "hhid": "Household ID",
    "s1q1": "Sex",
    "quintile": "Quintile",
    "s1q2": "Relation with the head of household",
    "s1q3y": "Age (Years)",
    "s1q4": "Marital status",
    "s1q5": "Place of birth",
    "s3q2": "Health insurance",
    "s4aq1": "Ever been to school",
    "s4aq3": "Highest school diploma or certificate obtained",
    "s4aq2": "Highest class succesfully attained in school",
    "s4aq7": "Attended school in the last 12 months",
    "s4bq4": "Able to read a simple written note",
    "s4bq5": "Able to write a simple letter",
    "s4bq6": "Able to do a written calculation",
    "s4bq8b": "Confidence to use computer",
    "s4bq7": "Has a mobile telephone"
})
```

```python
# Step 1: Derive household size (count of individuals per Household ID)
household_size = (
    person_features.groupby("Household ID")
    .size()
    .reset_index(name="Household size")
)

# Step 2: Derive number of kids (count of "Son or daughter of HH" per Household ID)
number_of_kids = (
    person_features[person_features["Relation with the head of household"] == "Son or daughter of HH"]
    .groupby("Household ID")
    .size()
    .reset_index(name="Number of kids")
)

# Step 3: Merge both into one dataframe
demographic = household_size.merge(number_of_kids, on="Household ID", how="left")

# Fill missing values for households with no kids
demographic["Number of kids"] = demographic["Number of kids"].fillna(0).astype(int)
```

```python
# Merge demographic info onto features_housing
features_housing = features_housing.merge(demographic, on="Household ID", how="left")
```

```python
# CROP

# Convert s7dq11 to numeric in case it's stored as a string or category
cs_s7d_large_crop["s7dq7"] = pd.to_numeric(cs_s7d_large_crop["s7dq7"], errors="coerce")

# Group by household and sum the harvest values
total_large_crop_value = (
    cs_s7d_large_crop
    .groupby("hhid")["s7dq7"]
    .sum()
    .reset_index()
    .rename(columns={
        "hhid": "Household ID",
        "s7dq7": "Total value of large crop harvest"
    })
)
```

```python
# Start from features_housing and merge the crop value info
final_large_crop_df = features_housing[["Household ID"]].merge(
    total_large_crop_value, on="Household ID", how="left"
)

# Create indicator column
final_large_crop_df["Household harvested any large crops in last 12 months"] = final_large_crop_df[
    "Total value of large crop harvest"
].notna().map({True: "yes", False: "no"})
```

```python
# Merge the large crop harvest info onto features_housing
features_housing = features_housing.merge(
    final_large_crop_df[["Household ID", "Total value of large crop harvest", "Household harvested any large crops in last 12 months"]],
    on="Household ID",
    how="left"
)
```

```python
# LIVESTOCK

# Convert s7aq6 to numeric
cs_s7a1_livestock["s7aq6"] = pd.to_numeric(cs_s7a1_livestock["s7aq6"], errors="coerce")

# Aggregate by Household ID
total_livestock_value = (
    cs_s7a1_livestock
    .groupby("hhid")["s7aq6"]
    .sum()
    .reset_index()
    .rename(columns={
        "hhid": "Household ID",
        "s7aq6": "Total value of livestock"
    })
)
```

```python
# Merge onto the full list of households
final_livestock_df = features_housing[["Household ID"]].merge(
    total_livestock_value, on="Household ID", how="left"
)

# Create the indicator column
final_livestock_df["Household raised any livestock in last 12 months"] = final_livestock_df[
    "Total value of livestock"
].notna().map({True: "yes", False: "no"})
```

```python
# Replace NaN with 0 first for clarity
final_livestock_df["Total value of livestock"] = final_livestock_df["Total value of livestock"].fillna(0)

# Now assign yes/no based on whether livestock value > 0
final_livestock_df["Household raised any livestock in last 12 months"] = final_livestock_df[
    "Total value of livestock"
].apply(lambda x: "yes" if x > 0 else "no")
```

```python
# Final merge onto the main features_housing dataframe
features_housing = features_housing.merge(
    final_livestock_df[["Household ID", "Total value of livestock", "Household raised any livestock in last 12 months"]],
    on="Household ID",
    how="left"
)
```

```python
# LAND (STAYS THE SAME)

# Convert s7cq4 to numeric (in case it's categorical or string)
cs_s7c_parcels["s7cq4"] = pd.to_numeric(cs_s7c_parcels["s7cq4"], errors="coerce")

# Group and sum by Household ID
total_land_area = (
    cs_s7c_parcels
    .groupby("hhid")["s7cq4"]
    .sum()
    .reset_index()
    .rename(columns={
        "hhid": "Household ID",
        "s7cq4": "Total surface area of land owned"
    })
)
```

```python
# Merge onto full list of households
final_land_df = features_housing[["Household ID"]].merge(
    total_land_area, on="Household ID", how="left"
)

# Replace NaN with 0
final_land_df["Total surface area of land owned"] = final_land_df["Total surface area of land owned"].fillna(0)

# Assign yes/no indicator
final_land_df["Household reported owning any land"] = final_land_df[
    "Total surface area of land owned"
].apply(lambda x: "yes" if x > 0 else "no")
```

```python
features_housing = features_housing.merge(
    final_land_df[["Household ID", "Total surface area of land owned", "Household reported owning any land"]],
    on="Household ID",
    how="left"
)
```

```python
# PERSON DATAFRAME (DROP EVERYONE EXCEPT HOUSEHOLD HEAD)

person_features = person_features[person_features["Relation with the head of household"] == "Household head _HH_"]
```

```python
features_housing = features_housing.merge(
    person_features,
    on="Household ID",
    how="left"
)
```

**Add the expenditure column**

```python
# Select relevant columns and rename hhid for merging
poverty_data = eicv5_poverty_file[["hhid", "sol_jan"]].rename(columns={"hhid": "Household ID"})

# Merge onto features_housing
features_housing = features_housing.merge(poverty_data, on="Household ID", how="left")
```

```python
# Rename that column to Equivalenced consumption
features_housing.rename(columns={"sol_jan": "Equivalenced consumption"}, inplace=True)
```

**Finishing touches**

```python
# Drop this column
features_housing = features_housing.drop(columns=["Relation with the head of household"], errors="ignore")
```
