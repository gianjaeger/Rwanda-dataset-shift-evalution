```python
# Select and rename relevant columns from vup_s0_s5_household
features_housing = vup_s0_s5_household[[
    "hhid",
    "province",
    "district",
    "ur2012",
    "ur2_2012",
    "weight",
    "clust",
    "s5aq1",
    "s5aq2",
    "s5aq5",
    "s5aq6",
    "s5aq10",
    "s5aq11",
    "s5cq1",
    "s5cq4",
    "s5cq11",
    "s5cq16",
    "s5cq17",
    "s5cq18",
    "s5cq19",
    "s5cq20",
    "s5cq25",
    "s5cq26",
    "s5dq1",
    "s5dq2",
    "s5dq3"
]].rename(columns={
    "hhid": "Household ID",
    "province": "Province",
    "district": "District (also stratum for SE calculation)",
    "ur2_2012": "Urban/rural 2002",
    "weight": "Final weight of household",
    "clust": "Cluster",
    "s5aq1": "Type of habitat",
    "s5aq2": "Type of dwelling",
    "s5aq5": "Number of rooms occupied by household",
    "s5aq6": "Number of bed rooms for household",
    "s5aq10": "Owner of the dwelling",
    "s5aq11": "Current occupancy status",
    "s5cq1": "Main source of drinking water",
    "s5cq4": "Distance to the nearest drinking water source (meters)",
    "s5cq11": "Amount paid to private water vendor last 7 days",
    "s5cq16": "Main source of home lighting",
    "s5cq17": "Amount paid on electricity in last 4 weeks",
    "s5cq18": "Primary source of cooking fuel",
    "s5cq19": "Mode of rubbish disposal",
    "s5cq20": "Type of toilet household use",
    "s5cq25": "Main source of info on environmental issues",
    "s5cq26": "Internet connection at home",
    "s5dq1": "Main construction material of exterior wall",
    "s5dq2": "Main roofing material for dwelling",
    "s5dq3": "Main floor material for dwelling"
})
```

```python
# Select and rename relevant columns
person_features = vup_s1_s2_s3_s4_s6a_s6e_s6f_person[[
    "hhid",
    "Consumption",
    "s1q1",
    "s1q2",
    "s1q3y",
    "s1q4",
    "s1q5",
    "s3q3",
    "s4aq1",
    "s4aq3",
    "s4aq2",
    "s4aq7",
    "s4bq3",
    "s4bq4",
    "s4bq5",
    "s4bq6",
    "s4bq8"
]].rename(columns={
    "hhid": "Household ID",
    "s1q1": "Sex",
    "s1q2": "Relation with the head of household",
    "s1q3y": "Age (Years)",
    "s1q4": "Marital status",
    "s1q5": "Place of birth",
    "s3q3": "Health insurance",
    "s4aq1": "Ever been to school",
    "s4aq3": "Highest school diploma or certificate obtained",
    "s4aq2": "Highest class succesfully attained in school",
    "s4aq7": "Attended school in the last 12 months",
    "s4bq3": "Able to read a simple written note",
    "s4bq4": "Able to write a simple letter",
    "s4bq5": "Able to do a written calculation",
    "s4bq6": "Confidence to use computer",
    "s4bq8": "Has a mobile telephone"
})
```

---

```python
# Step 1: Derive household size (count of individuals per Household ID)
household_size = (
    person_features.groupby("Household ID")
    .size()
    .reset_index(name="Household size")
)

# Step 2: Derive number of kids (count of "Son or daughter of HH" per Household ID)
number_of_kids = (
    person_features[person_features["Relation with the head of household"] == "Son or daughter of HH"]
    .groupby("Household ID")
    .size()
    .reset_index(name="Number of kids")
)

# Step 3: Merge both into one dataframe
demographic = household_size.merge(number_of_kids, on="Household ID", how="left")

# Fill missing values for households with no kids
demographic["Number of kids"] = demographic["Number of kids"].fillna(0).astype(int)
```

```python
# Merge demographic info onto features_housing
features_housing = features_housing.merge(demographic, on="Household ID", how="left")
```

```python
# Keep only household head rows
person_features = person_features[person_features["Relation with the head of household"] == "Household head (HH)"]
```

```python
# CROP

# Convert s7dq11 to numeric in case it's stored as a string or category
vup_s7d_large_crop["s7dq11"] = pd.to_numeric(vup_s7d_large_crop["s7dq11"], errors="coerce")

# Group by household and sum the harvest values
total_large_crop_value = (
    vup_s7d_large_crop
    .groupby("hhid")["s7dq11"]
    .sum()
    .reset_index()
    .rename(columns={
        "hhid": "Household ID",
        "s7dq11": "Total value of large crop harvest"
    })
)
```

```python
# Start from features_housing and merge the crop value info
final_large_crop_df = features_housing[["Household ID"]].merge(
    total_large_crop_value, on="Household ID", how="left"
)
```

```python
# Create indicator column
final_large_crop_df["Household harvested any large crops in last 12 months"] = final_large_crop_df[
    "Total value of large crop harvest"
].notna().map({True: "yes", False: "no"})
```

```python
# Merge the large crop harvest info onto features_housing
features_housing = features_housing.merge(
    final_large_crop_df[["Household ID", "Total value of large crop harvest", "Household harvested any large crops in last 12 months"]],
    on="Household ID",
    how="left"
)
```

```python
# LIVESTOCK

# Convert s7aq7 to numeric
vup_s7a1_livestock["s7aq7"] = pd.to_numeric(vup_s7a1_livestock["s7aq7"], errors="coerce")

# Aggregate by Household ID
total_livestock_value = (
    vup_s7a1_livestock
    .groupby("hhid")["s7aq7"]
    .sum()
    .reset_index()
    .rename(columns={
        "hhid": "Household ID",
        "s7aq7": "Total value of livestock"
    })
)
```

```python
# Merge onto the full list of households
final_livestock_df = features_housing[["Household ID"]].merge(
    total_livestock_value, on="Household ID", how="left"
)

# Create the indicator column
final_livestock_df["Household raised any livestock in last 12 months"] = final_livestock_df[
    "Total value of livestock"
].notna().map({True: "yes", False: "no"})
```

```python
# Replace NaN with 0 first for clarity
final_livestock_df["Total value of livestock"] = final_livestock_df["Total value of livestock"].fillna(0)

# Now assign yes/no based on whether livestock value > 0
final_livestock_df["Household raised any livestock in last 12 months"] = final_livestock_df[
    "Total value of livestock"
].apply(lambda x: "yes" if x > 0 else "no")
```

```python
# Final merge onto the main features_housing dataframe
features_housing = features_housing.merge(
    final_livestock_df[["Household ID", "Total value of livestock", "Household raised any livestock in last 12 months"]],
    on="Household ID",
    how="left"
)
```

```python
# LAND (STAYS THE SAME)

# Convert s7cq4 to numeric (in case it's categorical or string)
vup_s7c_parcels["s7cq4"] = pd.to_numeric(vup_s7c_parcels["s7cq4"], errors="coerce")

# Group and sum by Household ID
total_land_area = (
    vup_s7c_parcels
    .groupby("hhid")["s7cq4"]
    .sum()
    .reset_index()
    .rename(columns={
        "hhid": "Household ID",
        "s7cq4": "Total surface area of land owned"
    })
)
```

```python
# Merge onto full list of households
final_land_df = features_housing[["Household ID"]].merge(
    total_land_area, on="Household ID", how="left"
)

# Replace NaN with 0
final_land_df["Total surface area of land owned"] = final_land_df["Total surface area of land owned"].fillna(0)

# Assign yes/no indicator
final_land_df["Household reported owning any land"] = final_land_df[
    "Total surface area of land owned"
].apply(lambda x: "yes" if x > 0 else "no")
```

```python
features_housing = features_housing.merge(
    final_land_df[["Household ID", "Total surface area of land owned", "Household reported owning any land"]],
    on="Household ID",
    how="left"
)
```
