**Harmonize type of habitat**

```python
from pandas.api.types import CategoricalDtype

def harmonize_habitat_column(df, column_name="Type of habitat"):
    # Step 1: Define replacement map for standardization
    habitat_map = {
        "Umudugudu": "Umudugudu",
        "Unplanned clustered rural housing": "Unplanned clustered rural housing",
        "Isolated rural housing": "Isolated rural housing",
        "Unplanned urban housing": "Unplanned urban housing",
        "Modern planned area": "Modern planned area",
        "Aglomeration": "Aglomeration",
        "Small settlement": "Aglomeration",  # Merge with Aglomeration
        "Other": "Other"  # Only in 2010, optional to keep or drop
    }

    # Step 2: Define unified set of categories
    habitat_categories = [
        "Umudugudu",
        "Unplanned clustered rural housing",
        "Isolated rural housing",
        "Aglomeration",
        "Unplanned urban housing",
        "Modern planned area",
        "Other"
    ]

    # Step 3: Apply mapping and set categorical type
    df[column_name] = df[column_name].replace(habitat_map)
    df[column_name] = df[column_name].astype(CategoricalDtype(categories=habitat_categories, ordered=False))

    return df
```

```python
# Apply to each year’s dataframe
df2010 = harmonize_habitat_column(df2010)
df2013 = harmonize_habitat_column(df2013)
df2016 = harmonize_habitat_column(df2016)
```

```python
import numpy as np
import pandas as pd
from pandas.api.types import CategoricalDtype

def harmonize_diploma(df, column_name="Highest school diploma or certificate obtained"):
    df = df.copy()

    # ✅ Ensure it's not already categorical with 99.0 as a category
    df[column_name] = df[column_name].astype("object")

    # Replace coded missing values with np.nan
    df[column_name] = df[column_name].replace([99.0, "Missing"], np.nan)

    # Standardize text values
    diploma_map = {
        "License": "Professional license",
        "Baccalaureat _D6+2-3_": "Humanities Diploma"
    }
    df[column_name] = df[column_name].replace(diploma_map)

    # Define valid categories (make sure no nulls are included!)
    diploma_categories = [
        "Primary completed",
        "Post primary certificate",
        "Secondary common",
        "Diploma A3, D5, D4",
        "Humanities Diploma",
        "Professional license",
        "Bachelors",
        "Masters",
        "Doctorate",
        "Engineer",
        "No diploma"
    ]

    # Create and assign the categorical type
    cat_dtype = CategoricalDtype(categories=diploma_categories, ordered=True)
    df[column_name] = df[column_name].astype(cat_dtype)

    return df
```

```python
df2010 = harmonize_diploma(df2010)
df2013 = harmonize_diploma(df2013)
df2016 = harmonize_diploma(df2016)
```

```python
from pandas.api.types import CategoricalDtype

def harmonize_health_insurance(df, column_name="Health insurance"):
    df = df.copy()

    # Drop rows with invalid code 9.0
    df = df[df[column_name] != 9.0]

    # Define valid categories
    insurance_categories = [
        "None",
        "Mutual insurance",
        "MMI",
        "RAMA",
        "Employer",
        "Other insurance"
    ]

    # Standardize type
    cat_dtype = CategoricalDtype(categories=insurance_categories, ordered=False)
    df[column_name] = df[column_name].astype("object")
    df[column_name] = df[column_name].astype(cat_dtype)

    return df
```

```python
import pandas as pd
from pandas.api.types import CategoricalDtype

def harmonize_computer_confidence(df, column_name="Confidence to use computer"):
    df = df.copy()

    # Drop rows with invalid value 9.0
    df = df[df[column_name] != 9.0]

    # Harmonize naming conventions
    confidence_map = {
        "Yes": "Yes",
        "No": "No",
        "No literate": "Not literate",
        "Not literate": "Not literate"
    }

    df[column_name] = df[column_name].replace(confidence_map)

    # Define the final categorical order clearly
    confidence_categories = [
        "Not literate",
        "No",
        "Yes"
    ]

    # Assign categorical dtype (unordered here, but you could order if desired)
    cat_dtype = CategoricalDtype(categories=confidence_categories, ordered=True)
    df[column_name] = df[column_name].astype(cat_dtype)

    return df
```

```python
df2010 = harmonize_computer_confidence(df2010)
df2013 = harmonize_computer_confidence(df2013)
df2016 = harmonize_computer_confidence(df2016)
```
